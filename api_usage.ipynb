```python
import sys
sys.path.insert(0, r"C:\Users\e-malandaf\MUTLF-Ouyea")

import requests

url = "http://127.0.0.1:8000/predict"

frases = [
    "The company reported record-breaking profits this quarter, exceeding all analyst expectations.",
    "Shares plummeted after the CEO announced massive layoffs and a restructuring plan.",
    "The board approved the annual dividend payment as scheduled.",
    "Investors remain cautious as inflation data comes in higher than expected.",
    "Trading volume was average with no significant market movements today.",
]

print("=== FINANCIAL SENTIMENT ANALYSIS ===\n")
for frase in frases:
    respuesta = requests.post(url, json={"text": frase}).json()
    sentiment = respuesta["prediction"].upper()
    confianza = max(respuesta["probabilities"].values())
    print(f"TEXTO    : {frase}")
    print(f"SENTIMIENTO: {sentiment}  (confianza: {confianza:.1%})")
    print("-" * 70)
```
